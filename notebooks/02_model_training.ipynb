```python
#> --- Imports ---
import os
import numpy as np
import sys
from pathlib import Path
# --- Resolving Package Path ---
notebook_dir = Path.cwd()
src_path = (notebook_dir / ".." / "src").resolve()
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
# --- Importing API Files ---
from random_forest.forest import RandomForest
```

```python
# --- Importing Data ---
x_train = np.load(r"..\\data\\splits\\transactions\\x_train.npy", allow_pickle=True)
x_test = np.load(r"..\\data\\splits\\transactions\\x_test.npy", allow_pickle=True)
y_train = np.load(r"..\\data\\splits\\transactions\\y_train.npy", allow_pickle=True)
```

```python
# --- Getting an Instance from the Model ---
RandomForestModel = RandomForest()
# --- Creating list of Configs ---
n_trees_array =         [80,   15,   30,   30,   50,   30] #> n_trees
trees_max_depth_array = [50,   25,   30,   30,   30,   30] #> trees_max_depth
min_leaf_purity_array = [0.98, 0.90, 0.95, 0.90, 0.95, 0.98] #> min_leaf_purity
```

```python
# --- Training Loop ---
for i in range(len(n_trees_array)):
    RandomForestModel.fit(x_train, y_train, 
                          n_trees=n_trees_array[i], 
                          trees_max_depth=trees_max_depth_array[i], 
                          min_leaf_purity=min_leaf_purity_array[i])
    # --- Predicting ---
    predicted_y = RandomForestModel.predict(x_test, timer=False)
    # --- Saving The Predictions ---
    prediction_save_path = f"..\\artifacts\\predictions\\transactions\\config_{i+1}\\predicted_y"
    os.makedirs(prediction_save_path, exist_ok=True)
    np.save(prediction_save_path, predicted_y)
    # --- Saving the Model ---
    RandomForestModel.save_model(directory=f"..\\artifacts\\random_forest\\transactions\\config_{i+1}\\", silent_save=True)
```

```python
# --- Clean Up ---
del x_train, x_test, y_train
del RandomForestModel, predicted_y
```
